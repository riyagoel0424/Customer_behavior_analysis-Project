```python
## Checking Missing Values

df.isnull().sum()
```

Observation: Only the Review Rating column contains 37 missing values. All other columns are complete and do not contain missing data.

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))

# The dataset was grouped by Category, and missing Review Rating values were filled using the median rating of each category.
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

We standardized the column names by converting them to snake_case format so they are easier to use in Python, SQL, and Power BI without syntax issues.

Upper case-> lower case and space -> _ underscore

```python
# FEATURE ENGINEERING

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

## Creating Age Groups

A new feature called age_group was created using quartile-based segmentation.

Customers were divided into four age segments: Young Adult, Adult, Middle-aged, and Senior.

This makes customer behavior analysis easier and helps identify trends across different age groups.

```python
# CHECKING NEW AGE GROUP COLUMN

df[['age','age_group']].head(10)
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
```

## Creating Purchase Frequency Days Feature

The frequency_of_purchases column contained categorical values such as Weekly, Monthly, and Annually.

To enable numerical analysis, these categories were converted into their approximate equivalent number of days and stored in a new column called purchase_frequency_days.

```python
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
